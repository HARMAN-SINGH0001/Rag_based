```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
from youtube_transcript_api import YouTubeTranscriptApi
```

```python
video_id="ukzFI9rgwfU"
```

```python
ytt=YouTubeTranscriptApi()
```

```python
script_list=ytt.fetch(video_id,languages=['en'])
```

```python
script=" ".join(chunks.text for chunks in script)
```

```python
print(script_list)
```

```python
splitter=RecursiveCharacterTextSplitter(chunk_size=100,chunk_overlap=0)
split_list=splitter.create_documents([script])
```

```python
print(split_list)
```

```python
len(split_list)
```

```python
from langchain_ollama import OllamaEmbeddings
```

```python
from langchain_community.vectorstores import FAISS
```

```python
embedding=OllamaEmbeddings(model='nomic-embed-text')
```

```python
vector_store=FAISS.from_documents(split_list,embedding)
```

```python
print(vector_store)
```

```python
vector_store.index_to_docstore_id
```

```python
vector_store.get_by_ids(["8b8ff4d5-fcd5-4313-9f8c-b3560a1ba9e2"])
```

```python
retriever=vector_store.as_retriever(search_type='similarity',search_kwargs={'K':2})
```

```python
result=retriever.invoke("what is Machine Learning?")
```

```python
print(result)
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
```

```python
model=ChatOllama(model='tinyllama:chat')
```

```python
prompt=PromptTemplate(template="""You are a Helpfull Assistant,
                      provide the answer from the follwing context,
                      and if you have suffiecient knowldge,just honestly say i dont know
                      {context}
                      ,{question}""",
                    input_variables=["context","question"])
```

```python
question="is the toic of Machine Learning discussed? if yes then answer"
retriever_docs=retriever.invoke(question)
```

```python
context="/n/n".join(docs.page_content for docs in retriever_docs)
```

```python
final_prompt=prompt.invoke({'context':context,"question":question})
```

```python
final_prompt
```

```python
res=model.invoke(final_prompt)
```

```python
print(res.content)
```
