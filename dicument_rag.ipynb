```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
from langchain_core.output_parsers import StrOutputParser
```

```python
from langchain_community.document_loaders import TextLoader
```

```python
loader=TextLoader('E:/Rag_Based/mahatma.txt',encoding='utf=8')
```

```python
docs=loader.load()
```

```python
docs
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
splitter=RecursiveCharacterTextSplitter(
    chunk_size=30,
    chunk_overlap=5
)
```

```python
split_list=splitter.split_documents(docs)
```

```python
len(split_list)
```

```python
from langchain_community.vectorstores import FAISS
```

```python
from langchain_ollama import OllamaEmbeddings
embeddings=OllamaEmbeddings(model='nomic-embed-text')
```

```python
vector_store=FAISS.from_documents(split_list,embeddings)
```

```python
retriever=vector_store.as_retriever(search_type='similarity',search_kwargs={'k':4})
```

```python
answer=retriever.invoke('where was he live?')
```

```python
from langchain_ollama import ChatOllama
from langchain_core.prompts import PromptTemplate
```

```python
model=ChatOllama(model='tinyllama:chat')
```

```python
prompts=PromptTemplate(template="{context},{question}",
                    input_variables=['context','question'])
```

```python
question="how many simbling does he had"
retriever_docs=retriever.invoke(question)
```

```python
context="/n/n".join(docs.page_content for docs in retriever_docs )
```

```python
result=prompts.invoke({'context':context,'question':question})
```

```python
answer=model.invoke(result)
```
